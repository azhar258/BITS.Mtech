```python
import pandas as pd

```

```python
df = pd.read_csv("../../Data/housing_neha.csv")
```

```python
df.head()
```

```python
# Find the row with the maximum price
costliest = df.loc[df['price'].idxmax()]
print("Zip code with the costliest house:", costliest['zipcode'])
print("Costliest house value:", costliest['price'])
```

```python
num_grade_10 = df[df['grade'] == 10].shape[0]
print("Number of houses with grade 10:", num_grade_10)
```

```python
print("Total null values:", df.isnull().sum().sum())
```

```python
customer = df[df['id'] == 9126100861]
if not customer.empty:
    print("Waterfront status:", customer.iloc[0]['waterfront'])
else:
    print("Customer ID not found.")
```

```python
num_views_3 = df[df['view'] == 3].shape[0]
print("Number of houses with 3 views:", num_views_3)
```

```python
print("Lowest house price:", df['price'].min())
```

```python
largest = df.loc[df['sqft_living'].idxmax()]
print("Area with the biggest sqft_living:", largest['sqft_living'])
print("Corresponding zipcode:", largest['zipcode'])
```

```python
# Example function to apply to each row
def your_function_name(row):
	# For example, return price per sqft_living
	return row['price'] / row['sqft_living']

result = df.apply(your_function_name, axis=1)
print(result.head())
```

```python
def update_grade(row):
    if 0 < row['sqft_living'] <= 400:
        row['grade'] += 1
    elif row['sqft_living'] > 400:
        row['grade'] += 2
    return row

df = df.apply(update_grade, axis=1)
```
